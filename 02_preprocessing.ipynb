```python
import pandas as pd
import numpy as np
!pip install -q huggingface_hub
from huggingface_hub import hf_hub_download
from huggingface_hub import login
login()
```

```python
reviews_path = hf_hub_download(
    repo_id="Satyamkarn100/NeuraRec-Electronics",
    filename="raw/reviews.parquet",
    repo_type="dataset"
)

meta_path = hf_hub_download(
    repo_id = "Satyamkarn100/NeuraRec-Electronics",
    filename='raw/metadata.parquet',
    repo_type = 'dataset'
)
```

```python
reviews = pd.read_parquet(reviews_path)
print('Reviews Shape:',reviews.shape)
```

```python
reviews['datetime'] = pd.to_datetime(
    reviews['timestamp'],unit='ms'
)
```

```python
print(reviews['parent_asin'].nunique())
print()
reviews.head(10)
```

```python
meta = pd.read_parquet(meta_path)
print('Meta-Data Shape:',meta.shape)
meta.head()
```

```python
meta.head()
```

```python
meta['features'] = meta['features'].apply(
    lambda x: "" if x== [] else " ".join(map(str,x))
)
meta['description'] = meta['description'].apply(
    lambda x: "" if x==[] else " ".join(map(str,x))
)
```
