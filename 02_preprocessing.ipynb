```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
!pip install -q huggingface_hub
from huggingface_hub import hf_hub_download
from huggingface_hub import login
login()
```

```python
reviews_path = hf_hub_download(
    repo_id="Satyamkarn100/NeuraRec-Electronics",
    filename="raw/reviews.parquet",
    repo_type="dataset"
)
```

```python
reviews = pd.read_parquet(reviews_path)
print('Reviews Shape:',reviews.shape)
```

```python
reviews.head()
```

```python
reviews['datetime'] = pd.to_datetime(
    reviews['timestamp'],unit='ms'
)
```

```python
print("Unique Users:",reviews['user_id'].nunique())
print("Unique items:", reviews["parent_asin"].nunique())
print()
reviews.head(10)
```

```python
meta_paths = []

for i in range(1, 11):
    path = hf_hub_download(
        repo_id="Satyamkarn100/NeuraRec-Electronics",
        filename=f"raw/meta{i}.parquet",
        repo_type="dataset"
    )
    meta_paths.append(path)

print("Downloaded:", len(meta_paths), "files")
```

```python
meta = pd.concat(
    [pd.read_parquet(path) for path in meta_paths],
    ignore_index=True
)

print("Meta-Data Shape:", meta.shape)
meta.head()
```

```python
print("Metadata Shape:", meta.shape)
print("\nColumns:")
print(meta.columns.tolist())

print("\nData Types:")
print(meta.dtypes)

print("\nInfo:")
meta.info()
```

```python
meta.loc[meta["main_category"].isna(),["categories", "title"]].head(10)
```

```python
missing_mask_cat = meta['main_category'].isna()
meta.loc[missing_mask_cat,'main_category'] = meta.loc[
    missing_mask_cat,'categories'
].apply(lambda x: x[0]
        if isinstance(x,(list,tuple,np.ndarray)) and len(x)>0
        else np.nan)
```

```python
meta["main_category"].isna().sum()
```

```python
meta.loc[
    meta['store'].isna(),
 ['store','title','main_category','categories']].head(20)
```

```python
meta.loc[meta['store'].isna(),'title'].duplicated().sum()
```

```python
meta['features_key'] = meta['features'].apply(str)
meta['description_key'] = meta['description'].apply(str)
meta['categories_key'] = meta['categories'].apply(str)
```

```python
store_map = (
    meta[meta['store'].notna()]
    .groupby(['features_key', 'description_key', 'categories_key'])['store']
    .agg(lambda x: x.unique()[0] if x.nunique() == 1 else np.nan)
)
```

```python
meta.isna().sum()
```

```python
meta.to_parquet("/content/meta_clean.parquet", index=False)
reviews.to_parquet("/content/reviews_clean.parquet", index=False)
```

```python
from huggingface_hub import HfApi
api = HfApi()
repo_id = "Satyamkarn100/NeuraRec-Electronics"

api.upload_file(
    path_or_fileobj="/content/meta_clean.parquet",
    path_in_repo="processed/meta_clean.parquet",
    repo_id=repo_id,
    repo_type="dataset",
    commit_message="Save current cleaned metadata"
)

api.upload_file(
    path_or_fileobj="/content/reviews_clean.parquet",
    path_in_repo="processed/reviews_clean.parquet",
    repo_id=repo_id,
    repo_type="dataset",
    commit_message="Save current cleaned reviews"
)
```

## **Label Encoding**

```python
from sklearn.preprocessing import LabelEncoder
```

```python
user_encoder = LabelEncoder()
item_encoder = LabelEncoder()
reviews['user_idx'] = user_encoder.fit_transform(reviews['user_id'])
reviews['item_idx'] = item_encoder.fit_transform(reviews['parent_asin'])
```

```python
reviews[["parent_asin", "item_idx"]].head(10)
```

# **K-core verification**

```python
user_counts = reviews['user_idx'].value_counts()
item_counts = reviews['item_idx'].value_counts()
```

```python
print("Users with < 5 interactions:", (user_counts < 5).sum())
print("Items with < 5 interactions:", (item_counts < 5).sum())
```

```python
import joblib
joblib.dump(user_encoder, "/content/user_encoder.pkl")
joblib.dump(item_encoder, "/content/item_encoder.pkl")
```
