```python
meta_paths = []

for i in range(1, 11):
    path = hf_hub_download(
        repo_id="Satyamkarn100/NeuraRec-Electronics",
        filename=f"raw/meta{i}.parquet",
        repo_type="dataset"
    )
    meta_paths.append(path)

print("Downloaded:", len(meta_paths), "files")
```

```python
meta = pd.concat(
    [pd.read_parquet(path) for path in meta_paths],
    ignore_index=True
)

print("Meta-Data Shape:", meta.shape)
meta.head()
```

```python
print("Metadata Shape:", meta.shape)
print("\nColumns:")
print(meta.columns.tolist())

print("\nData Types:")
print(meta.dtypes)

print("\nInfo:")
meta.info()
```

```python
missing = pd.DataFrame({
    "Missing_Count": meta.isnull().sum(),
    "Missing_Percentage": (meta.isnull().sum() / len(meta) * 100).round(2)
})

missing.sort_values("Missing_Count", ascending=False)
```

```python
print("Price available:", meta["price"].notna().sum())
print("Price missing:", meta["price"].isna().sum())
```

### Missing Value Observations

- The `price` column has 67.30% missing values, making it unreliable for direct use without careful handling.
- `main_category` has 6.60% missing values, indicating some products lack category information.
- `store` has only 0.59% missing values and is mostly complete.
- Core recommendation-related fields such as `parent_asin`, `average_rating`, `rating_number`, `title`, `description`, and `categories` have no missing values.
- Since price is missing for most products, it should not be used as a primary feature unless an appropriate missing-value strategy is applied.

```python
meta['price'] = meta['price'].replace(
    ['None','none','NULL','null',''],np.nan
)

# Convert price to numeric
meta['price'] = pd.to_numeric(meta['price'], errors='coerce')

# Missing indicator
meta['price_missing'] = meta['price'].isna().astype(int)

before = meta['price'].isna().sum()
print("Missing before:", before)

# Same title based median imputation
title_median = meta.groupby('title')['price'].transform('median')

meta['price'] = meta['price'].fillna(title_median)

after = meta['price'].isna().sum()
print("Missing after title-match fill:", after)
print("Recovered rows:", before - after)
```

```python
computers = meta[
    meta["main_category"] == "Computers"
].copy()

computers["categories_group"] = computers["categories"].apply(
    lambda x: " > ".join(map(str, x))
    if isinstance(x, (list, tuple, np.ndarray))
    else str(x)
)
```

```python
computers_grouped = (
    computers
    .groupby("categories_group")["price"]
    .agg(
        total_products="count",
        unique_prices="nunique",
        min_price="min",
        median_price="median",
        max_price="max"
    )
    .reset_index()
)

display(computers_grouped)
```
