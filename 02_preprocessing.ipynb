```python
import pandas as pd
import numpy as np
!pip install -q huggingface_hub
from huggingface_hub import hf_hub_download
from huggingface_hub import login
login()
```

```python
reviews_path = hf_hub_download(
    repo_id="Satyamkarn100/NeuraRec-Electronics",
    filename="raw/reviews.parquet",
    repo_type="dataset"
)
```

```python
reviews = pd.read_parquet(reviews_path)
print('Reviews Shape:',reviews.shape)
```

```python
reviews.head()
```

```python
reviews['datetime'] = pd.to_datetime(
    reviews['timestamp'],unit='ms'
)
```

```python
print(reviews['parent_asin'].nunique())
print()
reviews.head(10)
```

```python
meta_paths = []

for i in range(1, 11):
    path = hf_hub_download(
        repo_id="Satyamkarn100/NeuraRec-Electronics",
        filename=f"raw/meta{i}.parquet",
        repo_type="dataset"
    )
    meta_paths.append(path)

print("Downloaded:", len(meta_paths), "files")
```

```python
meta = pd.concat(
    [pd.read_parquet(path) for path in meta_paths],
    ignore_index=True
)

print("Meta-Data Shape:", meta.shape)
meta.head()
```

```python
meta["price"] = pd.to_numeric(meta["price"], errors="coerce")
```

```python
def list_to_text(x):
  if x is None:
    return ""

  if isinstance(x,(list,np.ndarray)):
    if len(x) == 0:
      return ""
    return " ".join(map(str,x))

  return str(x)

meta["features"] = meta["features"].apply(list_to_text)
meta["description"] = meta["description"].apply(list_to_text)
```

```python
print(meta['features'].head().tolist())
print(meta['description'].head().tolist())
```

```python
print(type(meta['features'].iloc[0]))
print(type(meta['description'].iloc[0]))
```

```python
reviews_asins = set(reviews['parent_asin'])
meta_asins = set(meta['parent_asin'])

matched_asins = reviews_asins.intersection(meta_asins)
```

```python
print("Total review rows:", len(reviews))
print("Unique products in reviews:", len(reviews_asins))
print("Unique products in metadata:", len(meta_asins))
print("Matched products:", len(matched_asins))

coverage = len(matched_asins)/len(reviews_asins)*100
print("Meta-data Coverage:",round(coverage,2),'%')
```
